Q7. Sum of weights

```python
asia_cars = df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]]

asia_cars.head(7)
```

```python
X = asia_cars.head(7).values
print(X)
```

```python
XTX = X.T @ X
print(XTX)
```

```python
inv_XTX = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print(y)
```

```python
w = (inv_XTX @ X.T) @ y
print(w)
```

```python
w.sum()
```
